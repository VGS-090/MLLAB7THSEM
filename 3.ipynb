```python
# Program 3: Find-S Algorithm

# Dataset
data = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High", "Weak", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High", "Strong", "Cool", "Change", "Yes"]
]

# Number of attributes
num_attributes = len(data[0]) - 1

# Initialize the most specific hypothesis
hypothesis = ["0"] * num_attributes

print("Initial Hypothesis:")
print(hypothesis)

# Process each training example
for example in data:

    # Check whether the example is positive
    if example[-1] == "Yes":

        print("\nPositive Example:")
        print(example)

        # Compare each attribute
        for i in range(num_attributes):

            # If hypothesis has no value yet
            if hypothesis[i] == "0":
                hypothesis[i] = example[i]

            # If values are different, generalize to ?
            elif hypothesis[i] != example[i]:
                hypothesis[i] = "?"

        print("Updated Hypothesis:")
        print(hypothesis)

    else:
        # Find-S ignores negative examples
        print("\nNegative Example - Ignored")
        print(example)


# Display final hypothesis
print("\n================================")
print("Final Most Specific Hypothesis")
print("================================")

print(hypothesis)
```

```python
import pandas as pd

# -----------------------------------
# 1. Load Dataset
# -----------------------------------
df = pd.read_csv("playtennis.csv")

print("Original Dataset:")
print(df)

print("\nDataset Shape:", df.shape)
print("\nColumns:")
print(df.columns)


# -----------------------------------
# 2. Remove rows with missing values
# -----------------------------------
df = df.dropna().reset_index(drop=True)

print("\nDataset after removing missing values:")
print(df)

print("\nShape after removing missing values:", df.shape)


# -----------------------------------
# 3. Identify attributes and target
# -----------------------------------
target_column = "PlayTennis"

attributes = [
    "Outlook",
    "Temperature",
    "Humidity",
    "Wind"
]

print("\nTarget:", target_column)
print("Attributes:", attributes)


# -----------------------------------
# 4. Find-S Algorithm
# -----------------------------------
def find_s(data, attributes, target):

    # Select only positive examples
    positive_examples = data[
        data[target].str.lower() == "yes"
    ]

    print("\nNumber of Positive Examples:",
          len(positive_examples))

    # -----------------------------------
    # Initialize hypothesis with first
    # positive example
    # -----------------------------------
    first_positive = positive_examples.iloc[0]

    hypothesis = [
        first_positive[attribute]
        for attribute in attributes
    ]

    print("\nInitial Hypothesis:")
    print(hypothesis)

    print("\n-----------------------------------------")
    print("Find-S Iterations")
    print("-----------------------------------------")

    # -----------------------------------
    # Process every positive example
    # -----------------------------------
    for index, row in positive_examples.iterrows():

        old_hypothesis = hypothesis.copy()

        for i, attribute in enumerate(attributes):

            if hypothesis[i] != row[attribute]:
                hypothesis[i] = "?"

        # Print only when hypothesis changes
        if old_hypothesis != hypothesis:

            print(f"\nExample {index + 1}")
            print("Training Example:",
                  [row[attribute] for attribute in attributes])

            print("Updated Hypothesis:",
                  hypothesis)

    return hypothesis


# -----------------------------------
# 5. Run Find-S
# -----------------------------------
final_hypothesis = find_s(
    df,
    attributes,
    target_column
)


# -----------------------------------
# 6. Display Final Hypothesis
# -----------------------------------
print("\n=================================")
print("FINAL FIND-S HYPOTHESIS")
print("=================================")

for attribute, value in zip(
        attributes,
        final_hypothesis):

    print(f"{attribute} = {value}")
```
